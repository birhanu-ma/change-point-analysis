## C. Assumptions and Limitations

### Assumptions

The analysis is conducted under the following assumptions:

#### Market Efficiency Assumption
Brent oil prices are assumed to reflect publicly available information, including geopolitical events, economic shocks, and policy decisions, in a timely manner.

#### Event Timing Approximation
Event dates represent approximate start dates when market expectations are likely to have shifted. It is assumed that the impact of an event may begin slightly before or after the recorded date.

#### Distributional Assumptions
In the baseline Bayesian change point model, oil price changes (or log returns) are assumed to follow a normal distribution within each regime. This assumption simplifies the modeling process but may not fully capture extreme tail behavior.

#### Single Structural Change Focus
The core change point model assumes a limited number of dominant structural breaks over time, rather than continuous or overlapping regime changes.

#### Exogeneity of Events
External events (e.g., conflicts, sanctions, and OPEC decisions) are treated as exogenous to oil prices, meaning they are assumed not to be influenced by short-term price movements.

---

### Limitations

#### Correlation vs. Causation
The change point detection framework identifies points in time where the statistical properties of oil prices change. While detected change points may align temporally with known geopolitical or economic events, this analysis does **not** prove causal impact. The results indicate statistical correlation in time rather than definitive cause-and-effect relationships.

#### Overlapping Events
Multiple events may occur within close time periods, making it difficult to attribute a detected change point to a single event.

#### Model Simplification
The baseline model focuses primarily on changes in the mean level of prices. Other important dynamics—such as volatility clustering, asymmetric shocks, or nonlinear effects—are not explicitly modeled.

#### Omitted Variables
The analysis does not explicitly include other macroeconomic or financial variables (e.g., exchange rates, interest rates, or global GDP growth), which may also influence oil prices.

#### Data Quality and Granularity
Event data is based on publicly available sources and summarized timelines. Differences in reporting standards and timing may introduce measurement uncertainty.

## Understanding the Model and the Data

### Review of Key References

To ground the analysis in sound statistical and methodological principles, key references related to data science workflows, Bayesian inference, and change point analysis were reviewed. These references provide foundational understanding of:

- Time series characteristics such as non-stationarity and volatility clustering
- Bayesian modeling concepts, including prior specification and posterior inference
- Change point detection methods for identifying structural breaks in temporal data
- Monte Carlo Markov Chain (MCMC) techniques for parameter estimation and uncertainty quantification

The reviewed materials inform both the choice of modeling approach and the interpretation of results within the context of energy market dynamics.

---

### Analysis of Time Series Properties

Before applying any statistical model, the Brent oil price time series is examined to understand its fundamental properties.

#### Trend Analysis
Brent oil prices exhibit long-term trends and pronounced boom–bust cycles driven by global economic growth, technological changes, and geopolitical events. These trends indicate that the raw price series is non-stationary.

#### Stationarity Testing
Formal stationarity tests, such as the Augmented Dickey-Fuller (ADF) test, are used to assess whether the time series has a constant mean and variance over time. The results typically indicate non-stationarity in price levels, while transformations such as log returns tend to be closer to stationary.

#### Volatility Patterns
The data shows clear evidence of volatility clustering, where periods of high price fluctuations are followed by similar periods of elevated volatility. Such behavior is commonly observed during economic crises, conflicts, or major policy announcements.

---

### Implications for Modeling Choices

The observed time series properties directly inform the modeling strategy:

- **Non-stationarity** motivates the use of models that can handle structural changes rather than assuming constant parameters.
- **Volatility clustering** suggests that market regimes may shift during periods of stress or uncertainty.
- **Structural breaks** in trends and levels justify the application of change point detection models to identify regime shifts over time.

As a result, a Bayesian change point modeling framework is selected to capture these dynamics while explicitly accounting for uncertainty.

---

### Change Point Models: Purpose and Interpretation

Change point models are designed to identify points in time where the statistical properties of a time series change significantly. In the context of Brent oil prices, these changes may reflect shifts in market regimes caused by geopolitical conflicts, economic shocks, or policy interventions.

By allowing model parameters (such as the mean price level) to differ before and after a change point, these models help isolate periods where the underlying behavior of the market changes rather than focusing on short-term price noise.

The Bayesian formulation provides probabilistic estimates of both the timing and magnitude of these changes.

---

### Expected Outputs of the Change Point Analysis

The primary outputs of the Bayesian change point analysis include:

- **Posterior distribution of change point dates**, indicating when structural breaks are most likely to have occurred
- **Estimated parameter values before and after each change point**, such as changes in average price levels
- **Credible intervals**, which quantify uncertainty around all estimated parameters
- **Probabilistic statements**, allowing conclusions such as the likelihood that a change occurred around a specific date

---

### Limitations of Change Point Analysis

While change point models are powerful tools, they have important limitations:

- Detected change points indicate **statistical changes**, not definitive causes
- Closely spaced or overlapping events may be difficult to distinguish
- Model results depend on assumptions about distributions and the number of change points
- Simplified models may not fully capture complex market dynamics such as feedback loops or nonlinear effects

These limitations are acknowledged when interpreting results and forming conclusions.
