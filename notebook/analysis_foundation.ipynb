## A. Outline your analysis steps: 
Clearly document the steps and processes involved in analyzing the Brent oil prices data from data loading to insight generation.

## 1. Problem Definition

The objective of this analysis is to examine how major geopolitical, economic, and policy-related events have influenced Brent crude oil prices over time. Rather than focusing on short-term price fluctuations, the analysis emphasizes identifying **structural breaks**—periods where the underlying behavior of oil prices changes significantly. These structural changes are often associated with major global events such as political conflicts, economic crises, sanctions, and OPEC policy decisions.

---

## 2. Data Collection

Two primary data sources will be used:

- **Brent Oil Price Data**  
  Daily Brent crude oil prices recorded in USD per barrel, covering the period from May 20, 1987, to September 30, 2022.

- **External Event Dataset**  
  A curated dataset of major events relevant to the global oil market, including:
  - Geopolitical conflicts in oil-producing regions
  - OPEC production decisions
  - International economic sanctions
  - Global economic shocks and crises

Each event will be recorded with an approximate start date and a brief description.

---

## 3. Data Cleaning and Preparation

The raw data will be prepared to ensure consistency and suitability for time series modeling:

- Convert the `Date` column to a datetime format
- Handle missing price values by either:
  - Dropping missing observations, or
  - Applying forward-filling where justified
- Sort the dataset chronologically by date
- Compute **log returns** of oil prices to support stationarity analysis and reduce trend-related effects

---

## 4. Exploratory Data Analysis (EDA)

Exploratory analysis will be conducted to understand the key characteristics of the time series:

- Visualize long-term price trends and major market cycles
- Inspect volatility patterns and identify periods of high uncertainty
- Perform stationarity tests (e.g., Augmented Dickey-Fuller test)
- Analyze the distribution of prices and log returns

Insights from EDA will inform model assumptions and parameter choices.

---

## 5. Modeling Strategy

A **Bayesian Change Point Detection** approach will be applied to identify structural breaks in the time series:

- A mean-shift change point model will be used as the core analytical framework
- A latent change point variable will represent the time index at which price behavior changes
- Separate parameters will be estimated for periods before and after each detected change point

This approach allows uncertainty to be quantified and expressed probabilistically.

---

## 6. Interpretation and Event Mapping

Detected change points will be compared against the external event dataset:

- Temporal alignment between change points and known events will be examined
- Hypotheses will be formulated regarding which events may have contributed to observed shifts
- Results will be interpreted as **associations in time**, not definitive causal relationships

---

## 7. Communication of Results

Findings from the analysis will be communicated through multiple channels tailored to stakeholders:

- A written analytical report summarizing methodology and insights
- An interactive dashboard for exploring price trends, change points, and events
- Policy-oriented interpretations to support decision-making by investors, policymakers, and energy companies
